```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Load Dataset
df=pd.read_csv('weight_loss_dataset.csv')

# Input and Output
X=df[['Diet','Exercise','Sleep','Stress']].values
y=df['Weight_Loss'].values

# Build ANN Model
model=Sequential([
    tf.keras.Input(shape=(4,)),
    Dense(8,activation='relu'),
    Dense(8,activation='relu'),
    Dense(1,activation='sigmoid')
])

# Compile Model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train Model
model.fit(X,y,epochs=100,verbose=0)

# User Input
diet=float(input("Diet Score:"))
exercise=float(input("Exercise Hours:"))
sleep=float(input("Sleep Hours:"))
stress=float(input("Stress Level:"))

# Predict
user_data=np.array([[diet,exercise,sleep,stress]])
prediction=model.predict(user_data,verbose=0)[0][0]

# Convert Prediction to 0 or 1
prediction=int(prediction>=0.5)

if prediction==1:
    print("Likely to Achieve Weight Loss")
else:
    print("Unlikely to Achieve Weight Loss")
```
